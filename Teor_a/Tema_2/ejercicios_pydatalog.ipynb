```python
from pyDatalog import pyDatalog

pyDatalog.clear()

# Símbolos de predicado y variables
pyDatalog.create_terms(
    'libro,leido,genero,lector,aficionado_cf,aficionado_genero,'
    'X,L,G'
)

# HECHOS
+libro('dune')
+libro('1984')

+leido('ana','dune')
+leido('luis','1984')

+genero('dune','ciencia_ficcion')
+genero('1984','distopia')


# REGLAS

# Una persona es lectora si ha leído algún libro
lector(X) <= leido(X,L)

# Una persona es aficionada a la ciencia ficción
# si ha leído un libro de ciencia ficción
aficionado_cf(X) <= leido(X,L) & genero(L,'ciencia_ficcion')

# Una persona es aficionada a un género
# si ha leído algún libro de ese género
aficionado_genero(X,G) <= leido(X,L) & genero(L,G)


# CONSULTAS

print(bool(lector('ana')))
print(bool(lector('luis')))

print(bool(aficionado_cf('ana')))
print(bool(aficionado_cf('luis')))

print(bool(aficionado_genero('ana','ciencia_ficcion')))
print(bool(aficionado_genero('luis','distopia')))

# Todas las personas lectoras
print(lector(X))

# Todas las personas aficionadas y sus géneros
print(aficionado_genero(X,G))

```

```python
from pyDatalog import pyDatalog

pyDatalog.clear()

# ------------------------------------------------
# DECLARACIÓN DE PREDICADOS Y VARIABLES
# ------------------------------------------------

pyDatalog.create_terms(
    'conecta,pertenece,operativo,'
    'bus_conecta,puede_viajar,accesible_desde,'
    'B,L,O,D'
)


# ------------------------------------------------
# BASE DE CONOCIMIENTO: HECHOS
# ------------------------------------------------

# Líneas y localidades que conectan
+conecta('l1','benetusser','valencia')
+conecta('l2','catarroja','valencia')
+conecta('l3','valencia','paterna')

# Autobuses y líneas a las que pertenecen
+pertenece('b10','l1')
+pertenece('b20','l2')
+pertenece('b30','l3')

# Autobuses operativos
+operativo('b10')
+operativo('b30')


# ------------------------------------------------
# REGLAS
# ------------------------------------------------

# R1. Un autobús conecta dos localidades si pertenece
# a una línea que conecta esas localidades.
bus_conecta(B,O,D) <= pertenece(B,L) & conecta(L,O,D)


# R2. Un autobús permite viajar entre dos localidades
# si las conecta y además está operativo.
puede_viajar(B,O,D) <= bus_conecta(B,O,D) & operativo(B)


# R3. Una localidad es accesible desde otra si existe
# un autobús que permite realizar ese trayecto.
accesible_desde(O,D) <= puede_viajar(B,O,D)


# ------------------------------------------------
# CONSULTAS
# ------------------------------------------------

# 1. ¿B10 conecta Benetússer con Valencia?
print(bool(bus_conecta('b10','benetusser','valencia')))
# True


# 2. ¿B20 conecta Catarroja con Valencia?
print(bool(bus_conecta('b20','catarroja','valencia')))
# True


# 3. ¿Es posible viajar en B10 de Benetússer a Valencia?
print(bool(puede_viajar('b10','benetusser','valencia')))
# True


# 4. ¿Es posible viajar en B20 de Catarroja a Valencia?
print(bool(puede_viajar('b20','catarroja','valencia')))
# False


# 5. ¿Valencia es accesible desde Benetússer?
print(bool(accesible_desde('benetusser','valencia')))
# True


# 6. ¿Valencia es accesible desde Catarroja?
print(bool(accesible_desde('catarroja','valencia')))
# False


# 7. ¿Paterna es accesible desde Valencia?
print(bool(accesible_desde('valencia','paterna')))
# True


# 8. Todos los trayectos que pueden realizarse
# actualmente utilizando autobuses operativos
print(puede_viajar(B,O,D))
```

```python
#Ejercicio propuesto 1

from pyDatalog import pyDatalog

pyDatalog.clear()

# Predicados y variables
pyDatalog.create_terms(
    'bondadoso,rico,amigo,afecto,progenitor,ancestro,hereda,'
    'X,Y,Z'
)

# -------------------------
# HECHOS
# -------------------------

# H1. María es progenitora de Luis
+progenitor('maria','luis')

# H2. Luis es progenitor de Lucía
+progenitor('luis','lucia')

# H3. Pedro es rico
+rico('pedro')

# H4. María es rica
+rico('maria')

# H5. Lucía y Pedro son amigos
+amigo('lucia','pedro')

# H6. Lucía es bondadosa
+bondadoso('lucia')


# -------------------------
# REGLAS
# -------------------------

# R1. Si una persona bondadosa tiene afecto por otra
# persona rica, heredará de ella.
hereda(X,Y) <= bondadoso(X) & afecto(X,Y) & rico(Y)


# R2. Si una persona es amiga de otra,
# tiene afecto por ella.
afecto(X,Y) <= amigo(X,Y)


# R3. Si una persona es progenitor/a de otra,
# es su ancestro.
ancestro(X,Y) <= progenitor(X,Y)


# R4. Si una persona es ancestro de otra,
# esta última tiene afecto por su ancestro.
afecto(Y,X) <= ancestro(X,Y)


# R5. Si una persona es progenitor/a de otra
# y esta es progenitor/a de una tercera,
# la primera es ancestro de la tercera.
ancestro(X,Z) <= progenitor(X,Y) & progenitor(Y,Z)


# -------------------------
# CONSULTA
# -------------------------

print(bool(hereda('lucia','maria')))
```

```python
#Ejercicio propuesto 2

from pyDatalog import pyDatalog

pyDatalog.clear()

# Predicados y variables
pyDatalog.create_terms( 'region, frontera, color, pintada, conflicto, X, Y, C' )

+region('r1')
+region('r2')
+region('r3')
+region('r4')
+region('r5')

+frontera('r1','r2')
+frontera('r1','r3')
+frontera('r1','r4')
+frontera('r2','r4')
+frontera('r2','r5')
+frontera('r3','r4')
+frontera('r4','r5')

frontera(X,Y) <= (frontera(Y,X) & ~frontera(X,Y))

+color('rojo')
+color('amarillo')
+pintada('r1','rojo')
+pintada('r2','amarillo')

conflicto(X,C) <= (frontera(X,Y) & pintada(Y,C))
pintada(X,C) <= (region(X) & color(C) & ~conflicto(X,C))

print(bool(pintada('r3', 'amarillo')))
print(bool(pintada('r5', 'amarillo')))
```

```python
#Ejercicio propuesto 3

from pyDatalog import pyDatalog

pyDatalog.clear()

pyDatalog.create_terms(
    'bloque,valor,misma_fila,misma_columna,mismo_bloque,'
    'ocupada,conflicto,puede_poner,'
    'F,C,B,N,F2,C2'
)

# Definicion de bloques

nfilas = 4

# Bloque 1
for f in range(1, (nfilas//2)+1):
    for c in range(1, (nfilas//2)+1):
        +bloque(f, c, 'b1')

# Bloque 2
for f in range(1, (nfilas//2)+1):
    for c in range((nfilas//2)+1, nfilas+1):
        +bloque(f, c, 'b2')

# Bloque 3
for f in range((nfilas//2)+1, nfilas+1):
    for c in range(1, (nfilas//2)+1):
        +bloque(f, c, 'b3')

# Bloque 4
for f in range((nfilas//2)+1, nfilas+1):
    for c in range((nfilas//2)+1, nfilas+1):
        +bloque(f, c, 'b4')

print(bloque(F,C,B))

# VALORES INICIALES DEL SUDOKU
+valor(1,1,4)
+valor(2,3,3)
+valor(3,2,1)
+valor(4,4,2)

# CONFLICTO EN LA MISMA FILA
misma_fila(F,C,N) <= valor(F,C2,N)

# CONFLICTO EN LA MISMA COLUMNA
misma_columna(F,C,N) <= valor(F2,C,N)

# CONFLICTO EN EL MISMO BLOQUE
mismo_bloque(F,C,N) <= bloque(F,C,B) & bloque(F2,C2,B) & valor(F2,C2,N)

# CONFLICTO SI LA CASILLA ESTÁ OCUPADA
ocupada(F,C) <= valor(F,C,N)

# EXISTE CONFLICTO SI SE DA CUALQUIERA DE LAS TRES SITUACIONES
conflicto(F,C,N) <= misma_fila(F,C,N)
conflicto(F,C,N) <= misma_columna(F,C,N)
conflicto(F,C,N) <= mismo_bloque(F,C,N)
conflicto(F,C,N) <= ocupada(F,C)

# PODEMOS PONER UN NÚMERO SI NO HAY CONFLICTO
puede_poner(F,C,N) <= ~conflicto(F,C,N)

# ------------------------------------------------
# CONSULTAS
# ------------------------------------------------

print(bool(puede_poner(2,1,4)))
print(bool(puede_poner(2,1,1)))
print(bool(puede_poner(2,1,2)))
print(bool(puede_poner(2,1,3)))
print(bool(puede_poner(1,1,1)))
```
